```python
LAST_NAME = "ELIAS"
SEED_NUM = 7
FAVORITE_ARTIST = "TUBERO"

execution_log = []


def log_process(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        execution_log.append(f"{func.__name__} executed")
        return result
    return wrapper


@log_process
def generate_readings(last, seed, artist):
    base = [ord(c) for c in last[:3]] + [ord(c) for c in artist[:3]]
    readings = [(val + seed) % 200 for val in base]
    return readings


@log_process
def validate_readings(readings):
    results = []
    for r in readings:
        try:
            if 50 <= r <= 150:
                results.append((r, "Valid"))
            else:
                results.append((r, "Invalid"))
        except Exception:
            results.append((r, "Error"))
    return results


@log_process
def classify(results):
    valid_count = sum(1 for _, status in results if status == "Valid")
    invalid_count = sum(1 for _, status in results if status == "Invalid")
    condition = "Optimal" if valid_count > invalid_count else "Faulty"
    return valid_count, invalid_count, condition


readings = generate_readings(LAST_NAME, SEED_NUM, FAVORITE_ARTIST)
validation = validate_readings(readings)
valid_count, invalid_count, condition = classify(validation)


print("Generated Equipment Data:", readings)
print("Validation Results:", validation)
print("Diagnostic Results:",
      f"Valid={valid_count}, Invalid={invalid_count}, Condition={condition}")
print("Execution Log:", execution_log)
print("Final Output:",
      f"Out of {len(readings)} readings, {valid_count} valid, {invalid_count} invalid → {condition}")
```
