```python
LAST_NAME = "ELIAS"
SEED_NUM = 7
FAVORITE_ARTIST = "ELIAS"

execution_log = []

def generate_fault_code(last, seed, artist):
    # sum of ASCII values of initials * seed
    code = (ord(last[0]) + ord(artist[0])) * seed
    execution_log.append("generate_fault_code executed")
    return code

def fault_trace(code, calls=0, trace=None):
    if trace is None:
        trace = []
    trace.append(code)
    calls += 1
    # Base condition: stop when code < 50
    if code < 50:
        return trace, calls
    else:
        return fault_trace(code // 2, calls, trace)

fault_code = generate_fault_code(LAST_NAME, SEED_NUM, FAVORITE_ARTIST)
trace, call_count = fault_trace(fault_code)
execution_log.append("fault_trace executed")

print("Generated Fault Data:", fault_code)
print("Recursive Trace:", trace)
print("Number of Recursive Calls:", call_count)
print("Execution Log:", execution_log)
print("Final Output:",
      f"Fault traced in {call_count} steps → Final code {trace[-1]}")
```
